# Medical Insurance Cost Prediction

Predicting annual medical insurance cost from 10 health, lifestyle and demographic features (10,000 records). Eight regression models are compared, the best candidates are tuned with `GridSearchCV`, and the final model is saved for the Streamlit app (`app.py`).

# SCALING

```python
# The scaler is fitted on the TRAINING data only (see the next cell) to avoid data leakage
sc = StandardScaler()
```

## TRAIN TEST SPLIT

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=60)

# Fit the scaler on the training set only, then apply it to both sets
x_train = sc.fit_transform(x_train.values)
x_test = sc.transform(x_test.values)
```

# LINEAR REGRESSION

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
y_testpred = lr.predict(x_test)
y_trainpred = lr.predict(x_train)
mae = mean_absolute_error(y_test, y_testpred)
mse = mean_squared_error(y_test, y_testpred)
rmse = np.sqrt(mse)
lrt = r2_score(y_test, y_testpred) * 100
lrr = r2_score(y_train, y_trainpred) * 100
print("Linear Regression Test R2 Score:", lrt)
print("Linear Regression Train R2 Score:", lrr)
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
```

# DECISION TREE

```python
dt=DecisionTreeRegressor(random_state=42)
dt.fit(x_train,y_train)
y_testpred=dt.predict(x_test)
y_trainpred=dt.predict(x_train)
mae=mean_absolute_error(y_test,y_testpred)
mse=mean_squared_error(y_test, y_testpred)
dtt=r2_score(y_test,y_testpred)*100
dtr=r2_score(y_train,y_trainpred)*100
print('Test accuracy:',dtt)
print('Train accuracy:',dtr)
print("Mean absolute error:",mae)
print("Mean squared error:",mse)
```

# RANDOM FOREST REGRESSOR

```python
rf=RandomForestRegressor(random_state=42)
rf.fit(x_train,y_train)
y_testpred=rf.predict(x_test)
y_trainpred=rf.predict(x_train)
mae=mean_absolute_error(y_test,y_testpred)
mse=mean_squared_error(y_test, y_testpred)
rft=r2_score(y_test,y_testpred)*100
rfr=r2_score(y_train,y_trainpred)*100
print('Test accuracy:',rft)
print('Train accuracy:',rfr)
print("Mean absolute error:",mae)
print("Mean squared error:",mse)
```

# X-G BOOST

```python
xgb = XGBRegressor(random_state=42)
xgb.fit(x_train, y_train)
y_testpred=xgb.predict(x_test)
y_trainpred=xgb.predict(x_train)
mae=mean_absolute_error(y_test,y_testpred)
mse=mean_squared_error(y_test, y_testpred)
xgt=r2_score(y_test,y_testpred)*100
xgr=r2_score(y_train,y_trainpred)*100
print('Test accuracy:',xgt)
print('Train accuracy:',xgr)
print("Mean absolute error:",mae)
print("Mean squared error:",mse)
```

# SVR

```python
sv = SVR(kernel='rbf', C=100, gamma=0.1, epsilon=0.1)
sv.fit(x_train, y_train)
y_testpred = sv.predict(x_test)
y_trainpred = sv.predict(x_train)
mae = mean_absolute_error(y_test, y_testpred)
mse = mean_squared_error(y_test, y_testpred)
svt = r2_score(y_test, y_testpred) * 100
svr = r2_score(y_train, y_trainpred) * 100
print('Test accuracy:', svt)
print('Train accuracy:', svr)
print('Mean absolute error:', mae)
print('Mean squared error:', mse)
```

# KNN

```python
knn=KNeighborsRegressor(n_neighbors=3)
knn.fit(x_train,y_train)
y_testpred=knn.predict(x_test)
y_trainpred=knn.predict(x_train)
mae=mean_absolute_error(y_test,y_testpred)
mse=mean_squared_error(y_test, y_testpred)
knnt=r2_score(y_test,y_testpred)*100
knnr=r2_score(y_train,y_trainpred)*100
print('Test accuracy:',knnt)
print('Train accuracy:',knnr)
print("Mean absolute error:",mae)
print("Mean squared error:",mse)
```

# ADA BOOST

```python
adaboost=AdaBoostRegressor(random_state=42)
adaboost.fit(x_train,y_train)
y_testpred=adaboost.predict(x_test)
y_trainpred=adaboost.predict(x_train)
mae=mean_absolute_error(y_test,y_testpred)
mse=mean_squared_error(y_test, y_testpred)
adt=r2_score(y_test,y_testpred)*100
adr=r2_score(y_train,y_trainpred)*100
print('Test accuracy:',adt)
print('Train accuracy:',adr)
print("Mean absolute error:",mae)
print("Mean squared error:",mse)
```

# GradientBoosting

```python
gbc=GradientBoostingRegressor(random_state=42)
gbc.fit(x_train,y_train)
y_testpred=gbc.predict(x_test)
y_trainpred=gbc.predict(x_train)
mae=mean_absolute_error(y_test,y_testpred)
mse=mean_squared_error(y_test, y_testpred)
gbt=r2_score(y_test,y_testpred)*100
gbr=r2_score(y_train,y_trainpred)*100
print('Test accuracy:',gbt)
print('Train accuracy:',gbr)
print("Mean absolute error:",mae)
print("Mean squared error:",mse)
```

# HYPERPARAMETER TUNNING

```python
#Linear Regression
lr_params = {
    'fit_intercept': [True, False]
}
lr_grid = GridSearchCV(
    LinearRegression(),
    lr_params,
    cv=5
)
lr_grid.fit(x_train, y_train)
```

```python
# RANDOM FOREST
rf_params={'n_estimators':[10,50,100],'max_depth':[2,4,6,None]}
rf_grid=GridSearchCV(RandomForestRegressor(random_state=42),rf_params,cv=5)
rf_grid.fit(x_train,y_train)   
```

```python
# DECISION TREE
# ------------------
dt_params={'max_depth':[2,4,6,None]}
dt_grid=GridSearchCV(DecisionTreeRegressor(random_state=42),dt_params,cv=5)
dt_grid.fit(x_train,y_train)
```

```python
# KNN
knn_param = {
    'n_neighbors':range(1, 31),  # Number of neighbors
    'weights':['uniform','distance'],  # Voting strategy
    'metric':['euclidean','manhattan','minkowski']  # Distance metric
}
knn_grid= GridSearchCV(knn,knn_param,cv=5,scoring='r2', n_jobs=-1)
knn_grid.fit(x_train,y_train)
```

```python
# XGBoost
xgb_param = {
    'n_estimators': [100, 200, 300],
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [3, 5, 7],
    'min_child_weight': [1, 3, 5]
}
xgb_grid = GridSearchCV(xgb, xgb_param, cv=5, scoring='r2', n_jobs=-1)
xgb_grid.fit(x_train, y_train)
```

```python
models = {'Linear Regression': lr_grid, 'Random Forest': rf_grid, 'Decision Tree': dt_grid, 'KNN': knn_grid, 'XGB': xgb_grid}
for name, grid in models.items():
    best_model = grid.best_estimator_
    y_pred = best_model.predict(x_test)
    r2 = r2_score(y_test, y_pred)
    print(f"{name}:")
    print(f"  Best Params: {grid.best_params_}")
    print(f"  Test R2: {r2:.4f}\n")
```

```python
results = []
models = {'Linear Regression': lr_grid, 'Random Forest': rf_grid, 'Decision Tree': dt_grid, 'KNN': knn_grid, 'XGB': xgb_grid}
for name, grid in models.items():
    best_model = grid.best_estimator_
    results.append({
        'model': name,
        'test_r2': best_model.score(x_test, y_test) * 100,
        'train_r2': best_model.score(x_train, y_train) * 100
    })

tb = pd.DataFrame(results).sort_values('test_r2', ascending=False).reset_index(drop=True)
tb
```
